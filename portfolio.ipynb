```python
import json

# Charger le fichier de contenu
with open('content.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# Afficher le nom et le titre pour tester
print(f"Profil chargé : {data['profile']['name']} - {data['profile']['title']}")
print(f"Nombre de projets : {len(data['projects'])}")
```

```python
# Script de génération du site HTML à partir des données JSON
html_content = f"""<!DOCTYPE html>
<html lang="fr">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>{data['profile']['name']} - Portfolio</title>
    <style>
        body {{ font-family: Arial, sans-serif; max-width: 800px; margin: 40px auto; padding: 0 20px; line-height: 1.6; color: #333; }}
        header {{ text-align: center; margin-bottom: 40px; }}
        h1 {{ margin-bottom: 5px; }}
        .title {{ color: #555; font-weight: bold; }}
        .nav-links {{ margin: 20px 0; }}
        .nav-links a {{ margin: 0 10px; text-decoration: none; color: #0066cc; font-weight: bold; }}
        .nav-links a:hover {{ text-decoration: underline; }}
        .project {{ border: 1px solid #ddd; padding: 20px; border-radius: 8px; margin-bottom: 20px; background: #f9f9f9; }}
        .project h3 {{ margin-top: 0; }}
        .project a {{ display: inline-block; margin-top: 10px; color: #fff; background: #24292e; padding: 8px 15px; text-decoration: none; border-radius: 5px; }}
        .skills-list {{ display: flex; flex-wrap: wrap; gap: 8px; list-style: none; padding: 0; }}
        .skills-list li {{ background: #e1e4e8; padding: 5px 10px; border-radius: 4px; font-size: 14px; }}
    </style>
</head>
<body>

    <header>
        <h1>{data['profile']['name']}</h1>
        <p class="title">{data['profile']['title']}</p>
        <p>{data['profile']['bio']}</p>
        
        <div class="nav-links">
            <a href="{data['links']['linkedin']}" target="_blank">LinkedIn</a>
            <a href="{data['links']['github']}" target="_blank">GitHub</a>
            <a href="{data['links']['cv']}" target="_blank">Mon CV (PDF)</a>
            <a href="{data['links']['cover_letter']}" target="_blank">Lettre de motivation (PDF)</a>
            <a href="mailto:{data['links']['email']}">Contact</a>
        </div>
    </header>

    <section>
        <h2>Compétences Techniques</h2>
        <p><strong>Langages :</strong></p>
        <ul class="skills-list">
"""

# Ajouter dynamiquement les langages
for lang in data['skills']['languages']:
    html_content += f"            <li>{lang}</li>\n"

html_content += """        </ul>
        <p><strong>Frameworks & IA :</strong></p>
        <ul class="skills-list">
"""

# Ajouter dynamiquement les frameworks IA
for fw in data['skills']['frameworks_ai']:
    html_content += f"            <li>{fw}</li>\n"

html_content += """        </ul>
    </section>

    <section>
        <h2>Projets Phares</h2>
"""

# Ajouter dynamiquement chaque projet et son lien GitHub dédié
for proj in data['projects']:
    html_content += f"""
        <div class="project">
            <h3>{proj['title']}</h3>
            <p>{proj['description']}</p>
            <a href="{proj['github_url']}" target="_blank">Voir sur GitHub</a>
        </div>
    """

html_content += f"""
    </section>

    <footer>
        <hr style="margin: 40px 0; border: none; border-top: 1px solid #ddd;">
        <p style="text-align: center; color: #777; font-size: 14px;">© 2026 - {data['profile']['name']}</p>
    </footer>

</body>
</html>
"""

# Enregistrer le résultat dans un fichier index.html
with open('index.html', 'w', encoding='utf-8') as f:
    f.write(html_content)

print("Fichier 'index.html' généré avec succès depuis Jupyter !")
```
